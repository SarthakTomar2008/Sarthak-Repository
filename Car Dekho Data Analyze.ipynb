```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv(r"C:\Users\sarth\Downloads\Telegram Desktop\1776311302-P3-Car Market Trends Analysis with Car Dekho Data.csv")
df.head(5)
```

```python
df.describe()
```

```python

manual_cars = df[df["Transmission"].str.lower()=="manual"]
print("Manual cars present in the dataset: ",manual_cars)
```

```python
automatic_cars = df[df["Transmission"].str.lower()=="automatic"]
print("Automatic cars present in the data set: ",automatic_cars)
```

```python
transmission_counts = df["Transmission"].value_counts()

plt.Figure(figsize=(6, 6))
plt.pie(
    transmission_counts,
    labels=transmission_counts.index,
    autopct="%1.1f%%",
    startangle=90
)
plt.title("Manual vs Automatic Cars")
plt.show()
```

```python
petrol_cars = df[df["Fuel_Type"].str.lower()=="petrol"]
print("Petrol cars in the data set: ",petrol_cars)
```

```python
diesel_cars = df[df["Fuel_Type"].str.lower()=="diesel"]
print("Diesel cars in the dataset: ",diesel_cars)
```

```python
cng_cars = df[df["Fuel_Type"].str.lower()=="cng"]
print("CNG cars present in dataset: ",cng_cars)
```

```python
fuel_count = df["Fuel_Type"].value_counts()

plt.figure(figsize=(6, 6))
plt.pie(
    fuel_count,
    labels=fuel_count.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Diesel vs CNG vs Petrol")
plt.show()
```

```python
individual_cars = df[df["Seller_Type"].str.lower()=="individual"]
print("Cars through individual: ",individual_cars)
```

```python
dealer_cars = df[df["Seller_Type"].str.lower()=="dealer"]
print("Cars through dealer: ",dealer_cars)
```

```python
seller_count = df["Seller_Type"].value_counts()

plt.figure(figsize=(6, 6))
plt.pie(
    seller_count,
    labels=seller_count.index,
    autopct="%1.1f%%",
    startangle=90

)

plt.title("Dealer vs Individual")
plt.show()
```

```python

year_count = df["Year"].value_counts()

plt.figure(figsize=(6, 6))
plt.pie(
    year_count,
    labels=year_count.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Year Wise")
plt.show()
```
